## 2. Math: Changing the Variables

Python is basically a giant, super-fast calculator. You can add (`+`), subtract (`-`), multiply (`*`), and divide (`/`).

## 🔁 3. Loops: Doing Things Over and Over

Sometimes you want the computer to do something 100 times. You *could* type the instruction 100 times, but Vector is way too lazy for that.

Instead, we use a **Loop**. A `for` loop tells Python to repeat a block of code a certain number of times.
*(Notice how the code inside the loop is indented with spaces! That tells Python what belongs inside the loop).*

```python
print("Vector is getting ready for a nap...")

# This loop will repeat exactly 3 times
for i in range(3):
    print("Zzz... (snore)")

print("Vector woke up!")
```

## 🎮 5. Your First Interactive Lab: Vector's Energy Tracker

Let's put everything we learned together! We are going to build a simulation of Vector's Energy Level.

Here is the rule for our simulation:
* Vector starts with **10 Energy Points**.
* Every nap he takes adds **20 Energy Points**.
* Every snack he eats adds **5 Energy Points**.

We wrote the Python code to draw a graph of this rule. **Run the code block below, and play with the sliders to see how Naps and Snacks change Vector's total energy!**

```python
# This is a "Function". It is a recipe that Python saves for later.
def calculate_energy(naps, snacks):
    # The Math Rule
    base_energy = 10
    total_energy = base_energy + (naps * 20) + (snacks * 5)

    # ---------------------------------------------------------
    # Everything below here is just using 'plt' to draw the bar graph!
    # ---------------------------------------------------------

    fig, ax = plt.subplots(figsize=(6, 4))

    # Draw a bar chart with one giant bar
    ax.bar(["Vector's Energy"], [total_energy], color="#FF9A3C", width=0.5)

    # Draw a dotted line to show his "Hyperactive" zone
    ax.axhline(100, color="red", linestyle="--", label="Maximum Energy (Zoomies!)")

    # Make the graph look nice
    ax.set_ylim(0, 150)
    ax.set_ylabel("Energy Points", fontweight="bold")
    ax.set_title(f"Total Energy: {total_energy} Points!", fontweight="bold", fontsize=14)
    ax.legend()

    plt.show()

# This uses the 'widgets' toolbox to create sliders and connect them to our recipe!
widgets.interact(calculate_energy,
                 naps=widgets.IntSlider(min=0, max=5, step=1, value=2, description='Naps:'),
                 snacks=widgets.IntSlider(min=0, max=10, step=1, value=3, description='Snacks:'))
```
