## Simulation 1: The Delayed Shower

We wrote a Python function called `run_shower_sim`. It acts exactly like you stepping into a shower.
* **Target Temperature:** 100°F
* **Reaction Speed:** How aggressively you turn the knob when you feel the water is wrong.
* **Pipe Delay:** How many seconds it takes for the water to travel from the heater to your skin.

**Instructions:**
1. Run the code block below.
2. Play with the sliders.
3. Try setting the **Pipe Delay to 4**, and the **Reaction Speed to 1.5**. Look at the graph. Does the water ever settle down?

```python
def run_shower_sim(pipe_delay=3, reaction_speed=1.0):
    # Setup our shower environment
    target_temp = 100
    current_temp = 50 # Starts cold

    # We use a list to remember past temperatures (this is the pipe!)
    history = [current_temp] * 10

    # Lists to store our graph data
    times = []
    temperatures = []

    # Run the simulation for 30 seconds
    for second in range(30):
        times.append(second)
        temperatures.append(current_temp)

        # 1. SENSOR: You only feel the old water that just arrived
        felt_temp = history[-(pipe_delay + 1)]

        # 2. CONTROLLER: You decide how much to turn the knob
        error = target_temp - felt_temp
        knob_turn = error * reaction_speed

        # 3. ACTUATOR: The new water temp enters the pipes
        current_temp = current_temp + knob_turn
        history.append(current_temp)

    # Draw the graph
    plt.figure(figsize=(10, 5))
    plt.plot(times, temperatures, color='blue', linewidth=3, label='Actual Water Temp')
    plt.axhline(y=target_temp, color='green', linestyle='--', linewidth=2, label='Perfect Temp (100°F)')

    plt.title(f"Shower Temp over 30 Seconds (Delay: {pipe_delay}s, Reaction: {reaction_speed})", fontsize=14)
    plt.xlabel("Time (Seconds)", fontsize=12)
    plt.ylabel("Temperature (°F)", fontsize=12)
    plt.ylim(0, 200)
    plt.legend()
    plt.grid(True)
    plt.show()

# This creates the magic interactive sliders!
widgets.interact(run_shower_sim,
                 pipe_delay=widgets.IntSlider(min=0, max=8, step=1, value=3, description='Pipe Delay (s):'),
                 reaction_speed=widgets.FloatSlider(min=0.1, max=2.0, step=0.1, value=0.5, description='Reaction Speed:'))
```
