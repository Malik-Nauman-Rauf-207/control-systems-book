## Simulation 1: Vector's Moving Dot

A rate of change describes how much a quantity changes during a chosen amount of time.
Use the sliders below to set Vector's speed (Rate of Change) and the Time Passed. Watch how the math perfectly predicts his new position!

```python
def run_number_line(rate=3, t=2):
    start_pos = 8
    current_pos = start_pos + (rate * t)

    fig, ax = plt.subplots(figsize=(8, 2.5))

    # 1. Draw the number line grid
    ax.axhline(0, color='black', linewidth=2)
    ax.set_xlim(-20, 35)
    ax.set_ylim(-1, 1.5)
    ax.set_yticks([])

    # Draw tick marks
    for i in range(-20, 36, 5):
        ax.axvline(i, ymin=0.45, ymax=0.55, color='gray', zorder=1)
        ax.text(i, -0.4, str(i), ha='center', color='black', fontsize=10)

    # 2. Draw Start Position
    ax.plot(start_pos, 0, 's', color='gray', markersize=12, alpha=0.3, zorder=2)
    ax.text(start_pos, -0.8, "Start (8)", ha='center', color='gray', fontweight='bold')

    # 3. Draw Vector's Current Position
    ax.plot(current_pos, 0, 'o', color="#FF9A3C", markersize=18, zorder=4)
    ax.text(current_pos, 0.4, "Vector", ha='center', color="#FF9A3C", fontweight='bold', fontsize=12)

    # 4. Draw Arrow
    if current_pos != start_pos:
        arrow_color = '#2D8032' if rate > 0 else '#CC2222'
        ax.annotate('', xy=(current_pos, 0), xytext=(start_pos, 0),
                    arrowprops=dict(arrowstyle='-|>', color=arrow_color, lw=4, mutation_scale=20), zorder=3)

    ax.set_title(f"New Position = {start_pos} + ({rate} × {t})\nVector is at: {current_pos}", fontweight='bold', pad=15)
    for spine in ax.spines.values(): spine.set_visible(False)
    plt.show()

widgets.interact(run_number_line,
                 rate=widgets.IntSlider(min=-5, max=5, step=1, value=3, description='Rate (Speed):'),
                 t=widgets.IntSlider(min=0, max=5, step=1, value=2, description='Time (s):'))
```

## Simulation 2: The "Return to Zero" Walkway

What happens when the rule depends on where you already are?
The rule here is: `dx/dt = -x`.
This means the rate of change is the *negative* of your current position. Move the slider to place the mouse anywhere on the track and see how the arrow always forces it back to the balance point (0)!

```python
def run_return_to_zero(x=5):
    v = -x  # The Differential Equation!

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), gridspec_kw={'height_ratios': [1, 2.5]})

    # TOP PLOT
    ax1.axhline(0, color='black', linewidth=2)
    ax1.set_xlim(-12, 12); ax1.set_ylim(-1, 1.5); ax1.set_yticks([])
    for i in range(-10, 11, 2):
        ax1.axvline(i, ymin=0.45, ymax=0.55, color='gray', zorder=1)
        ax1.text(i, -0.4, str(i), ha='center', color='black', fontsize=10)

    ax1.plot(0, 0, 'x', color='blue', markersize=10, markeredgewidth=3, zorder=3)
    ax1.text(0, -0.8, "(0)", ha='center', color='blue', fontweight='bold')

    ax1.plot(x, 0, 'o', color="#FF9A3C", markersize=16, zorder=4)
    if x != 0:
        arrow_color = '#CC2222' if v < 0 else '#2D8032'
        ax1.annotate('', xy=(x + (v * 0.8), 0), xytext=(x, 0),
                     arrowprops=dict(arrowstyle='-|>', color=arrow_color, lw=4, mutation_scale=20), zorder=3)

    ax1.set_title("What happens on the walkway:", fontweight='bold')
    for spine in ax1.spines.values(): spine.set_visible(False)

    # BOTTOM PLOT
    x_vals = np.linspace(-11, 11, 100)
    ax2.plot(x_vals, -x_vals, color='gray', linestyle='--', linewidth=2, label='Rule: dx/dt = -x')
    ax2.axhline(0, color='black', lw=1); ax2.axvline(0, color='black', lw=1)

    ax2.plot(x, v, 'o', color="#FF9A3C", markersize=12, zorder=4)
    ax2.plot([x, x], [0, v], color="#FF9A3C", linestyle=':', lw=2)
    ax2.plot([0, x], [v, v], color="#FF9A3C", linestyle=':', lw=2)

    ax2.set_xlim(-11, 11); ax2.set_ylim(-11, 11)
    ax2.set_xlabel("Current Position (x)"); ax2.set_ylabel("Rate of Change (Speed)")
    ax2.set_title(f"Pos is {x}, Rate of Change is {v}", fontweight='bold')
    ax2.grid(True, linestyle=':', alpha=0.6); ax2.legend()
    plt.tight_layout(); plt.show()

widgets.interact(run_return_to_zero, x=widgets.IntSlider(min=-10, max=10, step=1, value=5, description='Position (x):'))
```

## Simulation 3: The Time Step Simulator (Euler's Method)

If the rate of change is constantly speeding up (`dW/dt = t`), taking large steps causes Vector's math to "fall behind" reality.
Adjust the Time Step (Δt) slider. Notice how smaller time steps make the simulation hug the true curve perfectly!

```python
def run_timestep_sim(dt=2.0):
    max_time = 10.0

    times_true = np.linspace(0, max_time, 100)
    water_true = 10 + 0.5 * times_true**2

    times_sim, water_sim = [0], [10]
    t, w = 0, 10

    while t < max_time:
        rate = t
        w = w + (rate * dt)
        t = t + dt
        if t > max_time:
            w = water_sim[-1] + (water_sim[-1] * (max_time - times_sim[-1]))
            t = max_time
        times_sim.append(t); water_sim.append(w)

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(times_true, water_true, color='#0077CC', linewidth=3, alpha=0.6, label="True Water Level")
    ax.plot(times_sim, water_sim, marker='o', color='#FF9A3C', linewidth=2.5, markersize=8, label=f"Vector's Math (Δt = {dt})")

    if dt > 0.1:
        ax.annotate('Error gap!', xy=(times_sim[-1], water_sim[-1]), xytext=(7, 40),
                    arrowprops=dict(facecolor='red', shrink=0.05, width=2, headwidth=8), color='red', fontweight='bold')

    ax.set_title(f"Why Step Size Matters (Δt = {dt} min)", fontweight='bold')
    ax.set_xlabel("Time (minutes)"); ax.set_ylabel("Water Amount (liters)")
    ax.set_xlim(-0.5, max_time + 0.5); ax.set_ylim(0, 70)
    ax.grid(True, linestyle=':', alpha=0.6); ax.legend(loc="upper left")
    for spine in ['top', 'right']: ax.spines[spine].set_visible(False)
    plt.show()

widgets.interact(run_timestep_sim, dt=widgets.FloatSlider(min=0.1, max=3.0, step=0.1, value=2.0, description='Time Step Δt:'))
```

## Simulation 4: The Shower Simulator

Vector's shower temperature changes based on how much heat the heater adds minus the heat lost to the room.
Adjust the Heater Power to see how a stronger heater changes the slope of the line over time!

```python
def run_shower_sim(heater=1.5):
    times = np.linspace(0, 20, 100)
    dt = times[1] - times[0]
    temperature = [60.0]

    for t in times[1:]:
        current_temp = temperature[-1]
        heat_loss = 0.1 * (current_temp - 60.0)
        rate_of_change = heater - heat_loss
        temperature.append(current_temp + (rate_of_change * dt))

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(times, temperature, color='#FF5733', linewidth=3, label='Shower Temperature')
    ax.axhline(100, color='red', linestyle='--', label='Scalding Danger Zone')
    ax.axhline(60, color='blue', linestyle=':', label='Cold Room Temp')

    ax.set_ylim(50, 110); ax.set_xlim(0, 20)
    ax.set_title("Simulating a Differential Equation", fontweight='bold')
    ax.set_xlabel("Time (seconds)"); ax.set_ylabel("Temperature (°F)")
    ax.legend(loc="upper left")
    plt.show()

widgets.interact(run_shower_sim, heater=widgets.FloatSlider(min=0.5, max=5.0, step=0.5, value=1.5, description='Heater Power:'))
```

## Simulation 5: The Soil Moisture Simulator

In this realistic greenhouse model, the water lost depends on how wet the soil already is (`dM/dt = u - kM`).
Notice how the moisture doesn't grow forever—it naturally curves and flattens out at an Equilibrium balance point!

```python
def run_soil_sim(u=5.0, k=0.2):
    equilibrium = u / k if k > 0 else 0
    max_time = 30; dt = 0.5
    times = np.arange(0, max_time + dt, dt)
    moisture = np.zeros_like(times)
    moisture[0] = 2.0

    for i in range(1, len(times)):
        dMdt = u - (k * moisture[i-1])
        moisture[i] = moisture[i-1] + (dMdt * dt)

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.axhline(equilibrium, color='green', linestyle='--', linewidth=2, label=f"Equilibrium ({equilibrium:.1f})")
    ax.plot(times, moisture, color='#0077CC', linewidth=3.5, label="Moisture Level")
    ax.plot(times[0], moisture[0], 'o', color='#FF9A3C', markersize=8, label="Start")

    ax.set_title(f"Simulating Soil Moisture (u={u}, k={k})", fontweight='bold')
    ax.set_xlabel("Time"); ax.set_ylabel("Moisture Level")
    ax.set_xlim(0, max_time); ax.set_ylim(0, max(50, equilibrium + 10))
    ax.grid(True, linestyle=':', alpha=0.6); ax.legend(loc="lower right")
    for spine in ['top', 'right']: ax.spines[spine].set_visible(False)
    plt.show()

widgets.interact(run_soil_sim,
                 u=widgets.FloatSlider(min=0.0, max=10.0, step=0.5, value=5.0, description='Pump (u):'),
                 k=widgets.FloatSlider(min=0.1, max=1.0, step=0.1, value=0.2, description='Loss (k):'))
```

## Simulation 6: The Overshoot Simulator (Rocket)

A controller that only looks at the angle ignores the rocket's momentum. See what happens when it pushes back toward 0 degrees.
Spoiler: The harder the Controller Strength (K), the more violently it oscillates!

```python
def run_rocket_sim(k=3.0):
    dt = 0.05
    max_time = 15.0
    times = np.arange(0, max_time, dt)
    angles, velocities = np.zeros_like(times), np.zeros_like(times)

    angles[0] = 30.0

    for i in range(1, len(times)):
        acceleration = -k * angles[i-1]
        velocities[i] = velocities[i-1] + (acceleration * dt)
        angles[i] = angles[i-1] + (velocities[i] * dt)

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.axhline(0, color='green', linestyle='--', linewidth=2, label="Perfectly Upright (0°)")
    ax.plot(times, angles, color='#CC2222', linewidth=3, label="Rocket Angle (θ)")

    ax.set_title(f"Rocket Angle over Time (Strength K={k})", fontweight='bold')
    ax.set_xlabel("Time (seconds)"); ax.set_ylabel("Tilt Angle (Degrees)")
    ax.set_xlim(0, max_time); ax.set_ylim(-60, 60)
    ax.grid(True, linestyle=':', alpha=0.6); ax.legend(loc="upper right")
    for spine in ['top', 'right']: ax.spines[spine].set_visible(False)
    plt.show()

widgets.interact(run_rocket_sim, k=widgets.FloatSlider(min=1.0, max=10.0, step=0.5, value=3.0, description='Strength (K):'))
```
